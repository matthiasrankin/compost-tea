```python
from datetime import datetime
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
START_MEAN = 5
START_STD = 1
SAMPLE_SIZE = 40

end_means_and_stds = {
    "control": {
        "mean": 10,
        "std": 1,
    },
    "treatment_1": {
        "mean": 12,
        "std": 0.9,
    },
    "treatment_2": {
        "mean": 13,
        "std": 0.8,
    },
    "treatment_3": {
        "mean": 15,
        "std": 0.8
    }
}
```

```python
group_dfs = []

for group, mean_and_std in end_means_and_stds.items():
    mean = mean_and_std["mean"]
    std = mean_and_std["std"]

    tmp_group_df = pd.DataFrame()
    tmp_group_df["group"] = [group] * SAMPLE_SIZE
    tmp_group_df["start_height"] = np.random.normal(START_MEAN, START_STD, size=SAMPLE_SIZE)
    tmp_group_df["end_height"] = np.random.normal(mean, std, size=SAMPLE_SIZE)
    group_dfs.append(tmp_group_df)

simulated_data = pd.concat(group_dfs, ignore_index=True)
simulated_data["plant_id"] = range(len(simulated_data))
```
