# Data Simulation

In the absence of real data, I'm randomly generating some with the assumptions that:

- Heights of plants follow a normal distribution.
- There is a somewhat linear, and statistically significant, relationship between the concentration of the compost tea and the growth of the plant.

I am making up figures like means, standard deviations, and concentrations, so these may be way off.

```python
from datetime import datetime
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
START_MEAN = 5
START_STD = 1
SAMPLE_SIZE = 40

group_params = {
    "control": {
        "mean": 10,
        "std": 1,
        "concentration": 0
    },
    "treatment_1": {
        "mean": 12,
        "std": 0.9,
        "concentration": 5,
    },
    "treatment_2": {
        "mean": 13,
        "std": 1,
        "concentration": 7.5,
    },
    "treatment_3": {
        "mean": 15,
        "std": 1.3,
        "concentration": 10
    }
}
```

```python
group_dfs = []

for group, params in group_params.items():
    mean = params["mean"]
    std = params["std"]
    concentration = params["concentration"]

    tmp_group_df = pd.DataFrame()
    tmp_group_df["group"] = [group] * SAMPLE_SIZE
    tmp_group_df["concentration"] = [concentration] * SAMPLE_SIZE
    tmp_group_df["start_height"] = np.random.normal(START_MEAN, START_STD, size=SAMPLE_SIZE)
    tmp_group_df["end_height"] = np.random.normal(mean, std, size=SAMPLE_SIZE)
    group_dfs.append(tmp_group_df)

simulated_data = pd.concat(group_dfs, ignore_index=True)

simulated_data["growth"] = simulated_data["end_height"] - simulated_data["start_height"]
simulated_data["plant_id"] = range(len(simulated_data))
simulated_data = simulated_data[["plant_id", "group", "concentration", "start_height", "end_height", "growth"]]
```
